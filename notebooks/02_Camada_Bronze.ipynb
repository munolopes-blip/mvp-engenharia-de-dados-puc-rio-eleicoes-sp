```python
from pyspark.sql import functions as F

# Caminho dos arquivos originais
BASE_PATH = "/Volumes/mvp_engenharia_de_dados_puc_rio/dados_tse/dados_originais"

print("Camada Bronze")
print("Origem:", BASE_PATH)
```

```python
display(dbutils.fs.ls(BASE_PATH))
```

```python
arquivo_2022 = f"{BASE_PATH}/consulta_cand_2022_SP.csv"

bronze_2022 = (
    spark.read.format("csv")
    .option("header", "true")
    .option("sep", ";")
    .option("encoding", "ISO-8859-1")
    .option("quote", '"')
    .option("inferSchema", "true")
    .load(arquivo_2022)
)

print("Registros carregados:", bronze_2022.count())
print("Colunas:", len(bronze_2022.columns))

display(bronze_2022.limit(10))
```

```python
bronze_2022.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "mvp_engenharia_de_dados_puc_rio.dados_tse.bronze_consulta_cand_2022"
    )

print("Tabela Bronze 2022 criada com sucesso.")
```

```python
display(
    spark.table(
        "mvp_engenharia_de_dados_puc_rio.dados_tse.bronze_consulta_cand_2022"
    ).limit(10)
)
```
