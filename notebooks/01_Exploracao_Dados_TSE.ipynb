```python
# Leitura exploratória do arquivo principal de candidaturas de 2022

arquivo_2022 = f"{BASE_PATH}/consulta_cand_2022_SP.csv"

df_2022 = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("sep", ";")
    .option("encoding", "ISO-8859-1")
    .option("quote", '"')
    .option("inferSchema", "true")
    .load(arquivo_2022)
)

print("Quantidade de registros:", df_2022.count())
print("Quantidade de colunas:", len(df_2022.columns))

display(df_2022.limit(10))
```

```python
# ============================================================
# Estrutura das colunas - TSE 2022
# ============================================================

print(f"Total de colunas: {len(df_2022.columns)}")
print("\nColunas e tipos:")

for i, field in enumerate(df_2022.schema.fields, start=1):
    print(f"{i:02d}. {field.name:<35} | {field.dataType}")
```

```python
# ============================================================
# Perfil inicial - cargos, CPF, gênero e raça
# ============================================================

print("========== CARGOS ==========")

display(
    df_2022
    .groupBy("CD_CARGO", "DS_CARGO")
    .count()
    .orderBy("CD_CARGO")
)


print("========== CPF ==========")

cpf_total = df_2022.count()

cpf_preenchido = (
    df_2022
    .filter(F.col("NR_CPF_CANDIDATO").isNotNull())
    .count()
)

cpf_nulo = cpf_total - cpf_preenchido

print(f"Total de registros: {cpf_total}")
print(f"CPF preenchido: {cpf_preenchido}")
print(f"CPF nulo: {cpf_nulo}")


print("========== GÊNERO ==========")

display(
    df_2022
    .groupBy("CD_GENERO", "DS_GENERO")
    .count()
    .orderBy("CD_GENERO")
)


print("========== COR / RAÇA ==========")

display(
    df_2022
    .groupBy("CD_COR_RACA", "DS_COR_RACA")
    .count()
    .orderBy("CD_COR_RACA")
)
```

```python
# ============================================================
# Recarregar os DataFrames principais
# ============================================================

from pyspark.sql import functions as F

BASE_PATH = "/Volumes/mvp_engenharia_de_dados_puc_rio/dados_tse/dados_originais"

# 2022
df_2022 = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("sep", ";")
    .option("encoding", "ISO-8859-1")
    .option("quote", '"')
    .option("inferSchema", "true")
    .load(f"{BASE_PATH}/consulta_cand_2022_SP.csv")
)

# 2026
df_2026 = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("sep", ";")
    .option("encoding", "ISO-8859-1")
    .option("quote", '"')
    .option("inferSchema", "true")
    .load(f"{BASE_PATH}/consulta_cand_2026_SP.csv")
)

print("DataFrames carregados com sucesso.")
print(f"2022: {df_2022.count()} registros")
print(f"2026: {df_2026.count()} registros")
```

```python
# ============================================================
# Validação da chave de identificação - CPF
# Universo: Deputado Estadual - SP
# ============================================================

df_2022_dep = df_2022.filter(F.col("CD_CARGO") == 7)
df_2026_dep = df_2026.filter(F.col("CD_CARGO") == 7)

print("========== 2022 ==========")

total_2022_dep = df_2022_dep.count()
cpf_distintos_2022 = (
    df_2022_dep
    .select("NR_CPF_CANDIDATO")
    .distinct()
    .count()
)

print(f"Registros: {total_2022_dep}")
print(f"CPFs distintos: {cpf_distintos_2022}")
print(f"CPFs duplicados: {total_2022_dep - cpf_distintos_2022}")


print("\n========== 2026 ==========")

total_2026_dep = df_2026_dep.count()
cpf_distintos_2026 = (
    df_2026_dep
    .select("NR_CPF_CANDIDATO")
    .distinct()
    .count()
)

print(f"Registros: {total_2026_dep}")
print(f"CPFs distintos: {cpf_distintos_2026}")
print(f"CPFs duplicados: {total_2026_dep - cpf_distintos_2026}")
```

```python
duplicados_2022 = (
    df_2022_dep
    .groupBy("NR_CPF_CANDIDATO")
    .count()
    .filter(F.col("count") > 1)
    .orderBy(F.desc("count"))
)

display(
    df_2022_dep
    .join(
        duplicados_2022.select("NR_CPF_CANDIDATO"),
        on="NR_CPF_CANDIDATO",
        how="inner"
    )
    .select(
        "NR_CPF_CANDIDATO",
        "SQ_CANDIDATO",
        "NR_CANDIDATO",
        "NM_CANDIDATO",
        "DS_GENERO",
        "DS_COR_RACA",
        "SG_PARTIDO",
        "NM_PARTIDO",
        "CD_SITUACAO_CANDIDATURA",
        "DS_SITUACAO_CANDIDATURA",
        "DS_SIT_TOT_TURNO"
    )
    .orderBy("NR_CPF_CANDIDATO", "SQ_CANDIDATO")
)
```

```python
print("========== SITUAÇÃO DA CANDIDATURA - 2022 ==========")

display(
    df_2022_dep
    .groupBy("CD_SITUACAO_CANDIDATURA", "DS_SITUACAO_CANDIDATURA")
    .count()
    .orderBy("CD_SITUACAO_CANDIDATURA")
)

print("========== SITUAÇÃO DA CANDIDATURA - 2026 ==========")

display(
    df_2026_dep
    .groupBy("CD_SITUACAO_CANDIDATURA", "DS_SITUACAO_CANDIDATURA")
    .count()
    .orderBy("CD_SITUACAO_CANDIDATURA")
)
```

```python
cpf_2022_aptos = (
    df_2022_dep
    .filter(F.col("CD_SITUACAO_CANDIDATURA") == 12)
    .select("NR_CPF_CANDIDATO")
    .distinct()
    .count()
)

cpf_2022_todos = (
    df_2022_dep
    .select("NR_CPF_CANDIDATO")
    .distinct()
    .count()
)

print("CPFs distintos - todos os registros 2022:", cpf_2022_todos)
print("CPFs distintos - somente APTO 2022:", cpf_2022_aptos)
```

```python
print("========== CPF NÃO DIVULGÁVEL ==========")

print("2022:")
display(
    df_2022_dep
    .groupBy("NR_CPF_CANDIDATO")
    .count()
    .filter(F.col("NR_CPF_CANDIDATO") < 0)
    .orderBy("NR_CPF_CANDIDATO")
)

print("2026:")
display(
    df_2026_dep
    .groupBy("NR_CPF_CANDIDATO")
    .count()
    .filter(F.col("NR_CPF_CANDIDATO") < 0)
    .orderBy("NR_CPF_CANDIDATO")
)
```

```python
# CPFs válidos de Deputado Estadual
cpfs_2022 = (
    df_2022_dep
    .filter(F.col("NR_CPF_CANDIDATO") > 0)
    .select("NR_CPF_CANDIDATO")
    .distinct()
)

cpfs_2026 = (
    df_2026_dep
    .filter(F.col("NR_CPF_CANDIDATO") > 0)
    .select("NR_CPF_CANDIDATO")
    .distinct()
)

# Candidatos presentes nos dois anos
comuns = cpfs_2022.join(
    cpfs_2026,
    on="NR_CPF_CANDIDATO",
    how="inner"
).count()

# Candidatos somente em 2026
somente_2026 = cpfs_2026.join(
    cpfs_2022,
    on="NR_CPF_CANDIDATO",
    how="left_anti"
).count()

# Candidatos somente em 2022
somente_2022 = cpfs_2022.join(
    cpfs_2026,
    on="NR_CPF_CANDIDATO",
    how="left_anti"
).count()

print("========== COMPARAÇÃO 2022 x 2026 ==========")
print("CPFs distintos válidos em 2022:", cpfs_2022.count())
print("CPFs distintos válidos em 2026:", cpfs_2026.count())
print("Presentes nos dois anos:", comuns)
print("Somente em 2026:", somente_2026)
print("Somente em 2022:", somente_2022)
```

```python
print("========== COR/RAÇA - DEPUTADO ESTADUAL ==========")

print("2022:")
display(
    df_2022_dep
    .groupBy("CD_COR_RACA", "DS_COR_RACA")
    .count()
    .orderBy("CD_COR_RACA")
)

print("2026:")
display(
    df_2026_dep
    .groupBy("CD_COR_RACA", "DS_COR_RACA")
    .count()
    .orderBy("CD_COR_RACA")
)
```

```python
id="9m3k7p"
print("========== COR/RAÇA — QUANTIDADE E % ==========")

raca_2022 = (
    df_2022_dep
    .groupBy("CD_COR_RACA", "DS_COR_RACA")
    .count()
    .withColumn(
        "percentual",
        F.round(F.col("count") / F.lit(df_2022_dep.count()) * 100, 2)
    )
    .withColumn("ano", F.lit(2022))
)

raca_2026 = (
    df_2026_dep
    .groupBy("CD_COR_RACA", "DS_COR_RACA")
    .count()
    .withColumn(
        "percentual",
        F.round(F.col("count") / F.lit(df_2026_dep.count()) * 100, 2)
    )
    .withColumn("ano", F.lit(2026))
)

resultado_raca = (
    raca_2022
    .unionByName(raca_2026)
    .select(
        "ano",
        "CD_COR_RACA",
        "DS_COR_RACA",
        "count",
        "percentual"
    )
    .orderBy("CD_COR_RACA", "ano")
)

display(resultado_raca)
```

```python
raca_comparacao = (
    resultado_raca
    .groupBy("CD_COR_RACA", "DS_COR_RACA")
    .pivot("ano", [2022, 2026])
    .agg(F.first("count"))
    .fillna(0)
    .withColumnRenamed("2022", "qtd_2022")
    .withColumnRenamed("2026", "qtd_2026")
    .withColumn(
        "variacao_qtd_pct",
        F.when(
            F.col("qtd_2022") > 0,
            F.round(
                (F.col("qtd_2026") - F.col("qtd_2022"))
                / F.col("qtd_2022") * 100,
                2
            )
        )
    )
    .orderBy("CD_COR_RACA")
)

display(raca_comparacao)
```

```python
print("========== GÊNERO — QUANTIDADE E % ==========")

genero_2022 = (
    df_2022_dep
    .groupBy("CD_GENERO", "DS_GENERO")
    .count()
    .withColumn(
        "percentual",
        F.round(
            F.col("count") / F.lit(df_2022_dep.count()) * 100,
            2
        )
    )
    .withColumn("ano", F.lit(2022))
)

genero_2026 = (
    df_2026_dep
    .groupBy("CD_GENERO", "DS_GENERO")
    .count()
    .withColumn(
        "percentual",
        F.round(
            F.col("count") / F.lit(df_2026_dep.count()) * 100,
            2
        )
    )
    .withColumn("ano", F.lit(2026))
)

resultado_genero = (
    genero_2022
    .unionByName(genero_2026)
    .select(
        "ano",
        "CD_GENERO",
        "DS_GENERO",
        "count",
        "percentual"
    )
    .orderBy("CD_GENERO", "ano")
)

display(resultado_genero)
```

```python
genero_comparacao = (
    resultado_genero
    .groupBy("CD_GENERO", "DS_GENERO")
    .pivot("ano", [2022, 2026])
    .agg(F.first("count"))
    .fillna(0)
    .withColumnRenamed("2022", "qtd_2022")
    .withColumnRenamed("2026", "qtd_2026")
    .withColumn(
        "variacao_qtd_pct",
        F.when(
            F.col("qtd_2022") > 0,
            F.round(
                (F.col("qtd_2026") - F.col("qtd_2022"))
                / F.col("qtd_2022") * 100,
                2
            )
        )
    )
    .orderBy("CD_GENERO")
)

display(genero_comparacao)
```
